```python
from pyspark.sql.functions import col
from pyspark.sql.types import *
control_path = (
    "abfss://customer-sales-datalake@nestlesalesdatalake.dfs.core.windows.net/"
    "metadata/pipeline_watermark/"
)
control_df = (
    spark.read
    .format("delta")
    .load(control_path)
)

gold_market_exists = (
    control_df
    .filter(
        col("pipeline_name") == "gold_market_sales"
    )
    .count() > 0
)

```

```python
control_schema = StructType([
    StructField("pipeline_name", StringType(), False),
    StructField("last_processed_timestamp", TimestampType(), True),
    StructField("last_run_id", StringType(), True),
    StructField("updated_at", TimestampType(), True)
])
```

```python
if gold_market_exists:
    print("gold_market_sales exists:", gold_market_exists)
else:
    gold_market_control_df = spark.createDataFrame(
    [
        (
            "gold_market_sales",
            None,
            None,
            None
        )
    ],
    control_schema
    )

    gold_market_control_df.write \
    .format("delta") \
    .mode("append") \
    .save(control_path)
```

```python
gold_market_control_df = (
    spark.read
    .format("delta")
    .load(control_path)
    .filter(
        col("pipeline_name") == "gold_market_sales"
    )
)

gold_market_last_processed_timestamp = (
    gold_market_control_df
    .select("last_processed_timestamp")
    .collect()[0][0]
)

print(
    "Gold Market last processed timestamp:",
    gold_market_last_processed_timestamp
)
```

```python
orders_path = (
    "abfss://customer-sales-datalake@nestlesalesdatalake.dfs.core.windows.net/"
    "silver/orders/"
)
orders_df = (
    spark.read
    .format("delta")
    .load(orders_path)
)

print("Silver Orders:", orders_df.count())

display(
    orders_df.limit(10)
)
```

```python
if gold_market_last_processed_timestamp is None:

    incremental_orders_df = orders_df

else:

    incremental_orders_df = (
        orders_df
        .filter(
            col("ingestion_timestamp") >
            gold_market_last_processed_timestamp
        )
    )

print(
    "Incremental orders:",
    incremental_orders_df.count()
)
```

```python
affected_markets_df = (
    incremental_orders_df
    .select("market")
    .distinct()
)

display(
    affected_markets_df
    .orderBy("market")
)
```

```python
affected_orders_df = (
    orders_df.alias("o")
    .join(
        affected_markets_df.alias("m"),
        col("o.market") == col("m.market"),
        "inner"
    )
    .select("o.*")
)

display(
    affected_orders_df
    .limit(20)
)
```

```python
from pyspark.sql.functions import sum, countDistinct, avg, round
affected_market_sales_df = (
    affected_orders_df
    .groupBy("market")
    .agg(
        sum("sales_amount").alias("total_sales"),
        countDistinct("order_id").alias("total_orders"),
        sum("quantity").alias("total_quantity"),
        round(
            avg("sales_amount"),
            2
        ).alias("average_order_value")
    )
)
```

```python
display(
    affected_market_sales_df
    .orderBy("market")
)
```

```python
gold_market_sales_path = (
    "abfss://customer-sales-datalake@nestlesalesdatalake.dfs.core.windows.net/"
    "gold/market_sales/"
)
```

```python
from delta.tables import DeltaTable

gold_market_table_exists = DeltaTable.isDeltaTable(
    spark,
    gold_market_sales_path
)

print(
    "Gold Market table exists:",
    gold_market_table_exists
)
```

```python
if not gold_market_table_exists:

    print("Gold Market table does not exist. Creating it.")

    (
        affected_market_sales_df
        .write
        .format("delta")
        .mode("overwrite")
        .save(gold_market_sales_path)
    )
else:
    print("Gold Market table exists. Running MERGE.")
    gold_market_table = DeltaTable.forPath(
        spark,
        gold_market_sales_path
    )
    (
        gold_market_table.alias("target")
        .merge(
            affected_market_sales_df.alias("source"),
            "target.market = source.market"
        )
        .whenMatchedUpdateAll()
        .whenNotMatchedInsertAll()
        .execute()
    )
```

```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    TimestampType
)

from pyspark.sql.functions import current_timestamp

gold_market_watermark_schema = StructType([
    StructField("pipeline_name", StringType(), False),
    StructField("last_processed_timestamp", TimestampType(), True),
    StructField("last_run_id", StringType(), True)
])

gold_market_watermark_df = (
    spark.createDataFrame(
        [
            (
                "gold_market_sales",
                gold_market_processed_timestamp,
                "GOLD_MARKET_SALES_RUN"
            )
        ],
        schema=gold_market_watermark_schema
    )
    .withColumn(
        "updated_at",
        current_timestamp()
    )
)

display(gold_market_watermark_df)
```

```python
control_table = DeltaTable.forPath(
    spark,
    control_path
)

(
    control_table.alias("target")
    .merge(
        gold_market_watermark_df.alias("source"),
        "target.pipeline_name = source.pipeline_name"
    )
    .whenMatchedUpdate(
        set={
            "last_processed_timestamp":
                "source.last_processed_timestamp",

            "last_run_id":
                "source.last_run_id",

            "updated_at":
                "source.updated_at"
        }
    )
    .whenNotMatchedInsert(
        values={
            "pipeline_name":
                "source.pipeline_name",

            "last_processed_timestamp":
                "source.last_processed_timestamp",

            "last_run_id":
                "source.last_run_id",

            "updated_at":
                "source.updated_at"
        }
    )
    .execute()
)
```
